```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from prophet import Prophet
```

```python
df=pd.read_csv("TRAFFICS (1).csv")
df=df[['DateTime','Vehicles']]
df.columns= ['ds','y']

df.to_csv("traffic_forecast.csv",index=False)
print("new file:TRAFFICS.csv")


print(" Model Training......")
model= Prophet(daily_seasonality=True,
               weekly_seasonality=True,
               yearly_seasonality=True)
df['ds']= pd.to_datetime(df['ds'],dayfirst=True)
model.fit(df)
print(" Model Training Completed")
```

```python
future= model.make_future_dataframe(periods=365*3,freq='H')
forecast=model.predict(future)

fig1 =model.plot(forecast)
plt.title('TRAFFIC FORECAST TILL 2027')
plt.show()

fig2= model.plot_components(forecast)
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Main Forecast Graph Save Karna
fig1 = model.plot(forecast)
plt.title('Traffic Forecast 2015-2027')
fig1.savefig('forecast_main.png', dpi=300, bbox_inches='tight')
print("Saved: forecast_main.png")

# 2. Components Graph Save Karna
fig2 = model.plot_components(forecast)
fig2.savefig('forecast_components.png', dpi=300, bbox_inches='tight')
print("Saved: forecast_components.png")
```
