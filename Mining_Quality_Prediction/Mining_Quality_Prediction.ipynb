```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.ensemble import RandomForestRegressor
```

```python
df = pd.read_csv('MiningProcess_Flotation_Plant_Database.csv', sep=',', decimal=',', on_bad_lines='skip',quotechar='"',engine='python')
df.columns = df.columns.str.strip()

df = df.iloc[:, 1:]
df = df.fillna(df.mean())

# df = df.sample(n=90000, random_state=42)
# print("Using rows:", df.shape)

target = '% Silica Concentrate'
print("Total Columns:", df.columns.tolist())
print("Shape:", df.shape)
```

```python
X1 = df.drop(target, axis=1)
y1 = df[target]

X_train, X_test, y_train, y_test = train_test_split(X1, y1, test_size=0.2, random_state=42)
model1 = RandomForestRegressor(n_estimators=20, random_state=42)
model1.fit(X_train, y_train)
r2_q1 = r2_score(y_test, model1.predict(X_test))
print("Q1 R2 Score:", round(r2_q1, 3))
```

```python
# ========= 60 MINUTES AHEAD PREDICTION =========
print("\n--- 60 MIN AHEAD PREDICTION ---")
df2 = df.copy()
df2['silica_future'] = df2[target].shift(-60)

df2['silica_rolling_mean_60'] = df2[target].rolling(window=60).mean()
df2 = df2.dropna()

X2 = df2.drop([target, 'silica_future'], axis=1)
y2 = df2['silica_future']

X_train, X_test, y_train, y_test = train_test_split(X2, y2, test_size=0.2, random_state=42)
model2 = RandomForestRegressor(n_estimators=30, random_state=42)
model2.fit(X_train, y_train)
r2_q2 = r2_score(y_test, model2.predict(X_test))
print("Q2 R2 Score:", round(r2_q2, 4))
```

```python
# ========= FINAL COMPARISON =========
print("\n--- FINAL RESULT ---")
result_df = pd.DataFrame({
    'Question': ['Q1 Live', 'Q2 60min Ahead'],
    'R2 Score': [round(r2_q1,3), round(r2_q2,3)]
})
print(result_df)
```
